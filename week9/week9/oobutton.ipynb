```python
import tkinter as tk

root = tk.Tk()
root.title("Oobutton")                
root.geometry("600x400")

lable = tk.Label(root, text = "객체지향윈도우프로그래밍")
lable.pack()

def on_btn1_clicked():
    print(text = "버튼을 클릭했군요!")

btn1 = tk.Button(root, text = "클릭해볼까요?", command = on_btn1_clicked)
btn1.pack()



root.mainloop()
```
